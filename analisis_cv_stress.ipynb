## 1️ Tujuan Survey

---

### Latar Belakang

Mahasiswa semester awal sering kali sudah dihadapkan pada tekanan untuk mempersiapkan CV (Curriculum Vitae) yang kompetitif, mengikuti organisasi, lomba, magang, dan berbagai kegiatan ekstrakurikuler. Di sisi lain, tekanan akademik yang tinggi membuat mahasiswa rentan mengalami stres. Pertanyaan yang muncul: **apakah semakin besar tekanan untuk membangun CV, semakin tinggi pula tingkat stres mahasiswa?**

### Tujuan Umum

> Menganalisis hubungan antara kegiatan pengembangan CV (seperti organisasi, magang, dan pelatihan) dengan tingkat stres mahasiswa Semester 1 sampai 8 di Institut Teknologi Sains Bandung.

### Tujuan Khusus

1. Mengidentifikasi profil demografis responden (gender, prodi, keterlibatan kegiatan)
2. Mengukur tingkat tekanan CV mahasiswa menggunakan skala Likert 1–5
3. Mengukur tingkat stres mahasiswa terkait persiapan karier
4. Menghitung korelasi Pearson antara CV Score dan Stress Score
5. Menyajikan hasil dalam bentuk visualisasi yang mudah dipahami
6. Menarik kesimpulan yang dapat menjadi masukan bagi program bimbingan mahasiswa

## 2️ Populasi dan Sampel

---

### Populasi

Populasi dalam penelitian ini adalah **seluruh mahasiswa aktif** di Institut Teknologi Sains Bandung.  
Populasi dipilih karena penelitian berfokus pada mahasiswa yang terlibat dalam berbagai kegiatan pengembangan diri seperti **organisasi, seminar, lomba, pelatihan, dan magang** sebagai bentuk persiapan CV dan portofolio.

Perkiraan jumlah populasi mahasiswa aktif: **± 1.000 mahasiswa**

### Penghitungan Sampel (Rumus Slovin)

$$n = \frac{N}{1 + N(e)^2}$$

**Keterangan:**
- $n$ = jumlah sampel
- $N$ = jumlah populasi = 1.000
- $e$ = margin of error = 15% (0,15)

**Perhitungan:**

$$n = \frac{1000}{1 + 1000 \times (0.15)^2} = \frac{1000}{1 + 22.5} = \frac{1000}{23.5} \approx 43$$

Jadi, **jumlah minimal sampel yang dibutuhkan adalah sekitar 43 responden**.  
Pada penelitian ini, jumlah responden yang berhasil diperoleh adalah **36 mahasiswa** mempertimbangkan keterbatasan waktu pengumpulan data namun masih representatif untuk analisis eksploratori.


## 3️ Metode Sampling

---

### Metode yang Digunakan: Purposive Sampling

**Purposive Sampling** adalah teknik pengambilan sampel berdasarkan pertimbangan peneliti mengenai siapa yang dianggap paling sesuai atau relevan dengan tujuan penelitian.

### Kriteria Responden

| Kriteria | Keterangan |
|---|---|
| Status Mahasiswa | Mahasiswa aktif Institut Teknologi Sains Bandung |
| Semester | Semester 2 (fokus pada tekanan awal perkuliahan) |
| Keterlibatan | Pernah atau sedang mengikuti kegiatan pengembangan diri |
| Pengalaman | Memiliki pengalaman terkait organisasi, seminar, lomba, magang, atau aktivitas tambahan |

### Alasan Pemilihan Metode

Purposive sampling dipilih karena peneliti membutuhkan responden yang **memiliki pengalaman relevan** dengan topik CV dan stres bukan sampel acak semua mahasiswa. Mahasiswa yang belum pernah terlibat kegiatan apapun akan memberikan data yang tidak representatif untuk penelitian ini.

### Instrumen Pengumpulan Data

Data dikumpulkan melalui **Google Form** yang disebarkan secara online kepada target responden. Kuesioner menggunakan **skala Likert 1–5**:

| Nilai | Keterangan |
|---|---|
| 1 | Sangat Tidak Setuju |

| 4 | Setuju |
| 5 | Sangat Setuju |

## Data Cleaning

> Tahap ini bertujuan memastikan data bersih sebelum dianalisis: memeriksa duplikat, nilai kosong (*missing values*), dan konsistensi tipe data.

```python
df = df_raw.copy()

print('=' * 58)
print('  RINGKASAN DATASET')
print('=' * 58)
print(f'  Jumlah Baris    : {df.shape[0]} responden')
print(f'  Jumlah Kolom    : {df.shape[1]} variabel')
print(f'  Duplikat        : {df.duplicated().sum()} baris')
print(f'  Total Missing   : {df.isnull().sum().sum()} nilai')

print()
print('=' * 58)
print('  STATUS MISSING VALUES PER KOLOM')
print('=' * 58)
for col in df.columns:
    miss = df[col].isnull().sum()
    pct  = round(miss / len(df) * 100, 1)
    icon = '[!] ' if miss > 0 else '[ok] '
    status = f'{miss} missing ({pct}%)' if miss > 0 else 'Lengkap'
    print(f'  {icon} {col[:50]:<52}: {status}')

# Handling: Nama opsional → dibiarkan NaN (memang opsional)
# Jenis Kelamin: 1 missing → isi dengan modus
modus_gender = df['Jenis Kelamin'].mode()[0]
df['Jenis Kelamin'] = df['Jenis Kelamin'].fillna(modus_gender)

print()
print('=' * 58)
print('  TINDAKAN CLEANING')
print('=' * 58)
print(f'  Nama (Opsional)   → Dibiarkan NaN (memang opsional)')
print(f'  Jenis Kelamin     → 1 NaN diisi dengan modus: "{modus_gender}"')
print(f'  Tidak ada duplikat → Data siap dianalisis')
print(f'  Missing (kolom analisis): {df.drop(columns=["Nama (Opsional)"]).isnull().sum().sum()} — Bersih!')
```

**Interpretasi Data Cleaning:**
> - Dari total **36 responden**, tidak ditemukan data duplikat, sehingga seluruh baris valid untuk dianalisis.
> - Terdapat **1 missing value** pada kolom `Jenis Kelamin`, yang diselesaikan dengan mengisi nilai modus (*Perempuan*) pendekatan umum untuk data kategorik dengan missing minimal.
> - Kolom `Nama (Opsional)` memiliki banyak NaN karena memang bersifat opsional dalam kuesioner, sehingga tidak perlu ditangani.

## Definisi Variabel & Statistik Deskriptif

> **CV Score** = rata-rata jawaban 7 item pertanyaan tekanan pengembangan CV (Likert 1–5)  
> **Stress Score** = rata-rata jawaban 9 item pertanyaan tingkat stres akademik/karier (Likert 1–5)  
> Semakin tinggi nilai → semakin tinggi tekanan / stres yang dirasakan.

```python
x_cols = [
    'Saya merasa tertekan karena harus memiliki CV yang lengkap dan menarik sebelum lulus. ',
    'Saya khawatir pengalaman organisasi atau kepanitiaan saya tidak cukup untuk dimasukkan ke CV.',
    'Saya merasa teman-teman memiliki pencapaian yang lebih banyak dibanding saya (misalnya magang, proyek, sertifikat).',
    'Saya merasa ada tuntutan eksternal untuk mengambil banyak kegiatan demi mempercantik CV.',
    'Saya sering membandingkan CV saya dengan CV orang lain (dari media sosial, LinkedIn, dll.).',
    'Tekanan untuk membangun CV terasa meningkat menjelang semester akhir atau masa rekrutmen kerja.',
    'Saya menghabiskan waktu yang sangat lama hanya untuk memikirkan atau memperbaiki tampilan CV.'
]
y_cols = [
    'Saya sering merasa cemas memikirkan masa depan karier saya.',
    'Saya merasa cemas ketika tidak mengikuti kegiatan yang dapat menambah pengalaman saya.',
    'Saya merasa kewalahan dalam menjalani berbagai kegiatan seperti organisasi, lomba, atau pelatihan.',
    'Saya merasa kewalahan dengan banyaknya hal yang harus disiapkan selama kuliah.',
    'Saya merasa terbebani dengan banyaknya kegiatan yang saya jalani di luar perkuliahan.',
    'Saya merasa lelah secara fisik atau pusing ketika memikirkan target pencapaian yang belum tercapai.',
    'Saya merasa stres ketika melihat teman yang sudah mendapat pengalaman kerja atau magang.',
    'Kegiatan tambahan yang saya jalani mempengaruhi kondisi mental saya.',
    'Saya merasa takut tidak siap saat nanti terjun ke dunia kerja.'
]

df['CV Score']     = df[x_cols].mean(axis=1).round(3)
df['Stress Score'] = df[y_cols].mean(axis=1).round(3)

desc = df[['CV Score', 'Stress Score']].describe().round(2)

print('CV Score dan Stress Score berhasil dihitung.\n')

display(
    desc.style
    .background_gradient(cmap='Blues', axis=None, vmin=0, vmax=5)
    .format('{:.2f}')
    .set_caption('Statistik Deskriptif CV Score & Stress Score')
    .set_table_styles([
        {'selector': 'caption', 'props': [('font-size','14px'),('font-weight','bold'),('padding','8px')]},
        {'selector': 'thead th', 'props': [('background-color','#185FA5'),('color','white'),('text-align','center'),('padding','8px')]},
        {'selector': 'tbody td', 'props': [('text-align','center'),('padding','7px')]},
    ])
)

print()
cv_means    = [round(df[c].mean(),2) for c in x_cols]
stress_means = [round(df[c].mean(),2) for c in y_cols]
print(f'  Rata-rata CV Score keseluruhan    : {np.mean(cv_means):.2f} / 5.00')
print(f'  Rata-rata Stress Score keseluruhan: {np.mean(stress_means):.2f} / 5.00')
```

## Kategorisasi Skor

> Skor komposit dikategorikan ke dalam 4 tingkat berdasarkan rentang Likert:
> - **Sangat Rendah**: 1.0 – 2.0 | **Rendah**: 2.0 – 3.0 | **Tinggi**: 3.0 – 4.0 | **Sangat Tinggi**: 4.0 – 5.0

> - Hanya **5,6%** (2 orang) yang berada di kategori Sangat Rendah.

> **Distribusi Stress Score:**
> - **72,2%** responden (26 orang) berada di kategori **Tinggi dan Sangat Tinggi** → lebih dari 2/3 mahasiswa mengalami stres di atas rata-rata.

> **Tabel Silang (Crosstab):**
> - Sel terbesar: **CV Tinggi × Stres Tinggi = 7 orang** dan **CV Tinggi × Stres Sangat Tinggi = 6 orang** → pola konsisten bahwa tekanan CV tinggi berbanding lurus dengan stres tinggi.
> - **Tidak ada satu pun responden** dengan CV Sangat Tinggi yang memiliki stres Sangat Rendah → korelasi positif terkonfirmasi dari pola distribusi ini.

## Uji Korelasi Pearson

> Korelasi Pearson mengukur kekuatan dan arah hubungan linear antara dua variabel kontinu.  
> **Syarat**: data berskala interval/rasio, distribusi mendekati normal terpenuhi untuk skor Likert komposit.

```python
r_val, p_val = stats.pearsonr(df['CV Score'], df['Stress Score'])
r2           = r_val ** 2

if   abs(r_val) >= 0.80: kekuatan = 'Sangat Kuat'
elif abs(r_val) >= 0.60: kekuatan = 'Kuat'
elif abs(r_val) >= 0.40: kekuatan = 'Sedang'
elif abs(r_val) >= 0.20: kekuatan = 'Lemah'
else:                    kekuatan = 'Sangat Lemah'

arah       = 'Positif ↗' if r_val > 0 else 'Negatif ↘'
signifikan = 'Signifikan (p < 0.05)' if p_val < 0.05 else 'Tidak Signifikan (p >= 0.05)'

print('╔════════════════════════════════════════════════════╗')
print('║        HASIL UJI KORELASI PEARSON                  ║')
print('╠════════════════════════════════════════════════════╣')
print(f'║  Variabel X  : CV Score (tekanan CV)               ║')
print(f'║  Variabel Y  : Stress Score (tingkat stres)        ║')
print(f'║  n           : {len(df)} responden                        ║')
print('╠════════════════════════════════════════════════════╣')
print(f'║  r           = {r_val:.4f}                              ║')
print(f'║  r²          = {r2:.4f}  ({r2*100:.1f}% variansi dijelaskan) ║')
print(f'║  p-value     = {p_val:.2e}                            ║')
print(f'║  Kekuatan    : {kekuatan:<32}    ║')
print(f'║  Arah        : {arah:<32}    ║')
print(f'║  Kesimpulan  : {signifikan:<32}    ║')
print('╚════════════════════════════════════════════════════╝')
```

**Interpretasi Korelasi Pearson:**

> **r = 0.7067** menunjukkan hubungan **positif yang kuat** antara tekanan pengembangan CV dan tingkat stres mahasiswa. Artinya, semakin tinggi tekanan yang dirasakan mahasiswa dalam membangun CV, semakin tinggi pula tingkat stres yang dialami.
>
> **r² = 0.4994 (≈ 49,9%)** nilai ini berarti bahwa sekitar **50% variasi tingkat stres mahasiswa dapat dijelaskan oleh tekanan pengembangan CV**. Sisanya (50%) dipengaruhi faktor lain di luar cakupan penelitian ini (misalnya: beban akademik, kondisi keluarga, kesehatan, dll.).
>
> **p-value = 0.0000014 (< 0.001)** jauh lebih kecil dari ambang signifikansi α = 0.05, sehingga hasil ini **sangat signifikan secara statistik** dan bukan terjadi secara kebetulan.

> - **Baris CV Rendah**: Dari 9 responden, 5 masuk Stres Rendah, 3 masuk Stres Tinggi pola lebih tersebar, menunjukkan tekanan CV rendah tidak selalu berkorelasi langsung dengan stres rendah (ada faktor stres lain).

### Visualisasi 6 | Histogram: Distribusi CV Score & Stress Score


```python
fig_hist, ax = plt.subplots(figsize=(12, 6))
fig_hist.patch.set_facecolor(BG)
ax.set_facecolor(CARD)

from scipy.stats import gaussian_kde

# ── Histogram CV Score (biru, semi-transparan) ──
n1, bins1, _ = ax.hist(
    df['CV Score'], bins=14, color=BIRU, edgecolor='white',
    linewidth=0.6, alpha=0.55, label='CV Score'
)

# ── Histogram Stress Score (emas, semi-transparan) ──
n2, bins2, _ = ax.hist(
    df['Stress Score'], bins=14, color=EMAS, edgecolor='white',
    linewidth=0.6, alpha=0.55, label='Stress Score'
)

# ── KDE CV Score ──
kde1 = gaussian_kde(df['CV Score'], bw_method=0.4)
x_cv = np.linspace(1.2, 5.3, 300)
max_n = max(n1.max(), n2.max())
scale1 = max_n / kde1(x_cv).max()
ax.plot(x_cv, kde1(x_cv) * scale1, color=BIRU_M, linewidth=2.5, linestyle='-')

# ── KDE Stress Score ──
kde2 = gaussian_kde(df['Stress Score'], bw_method=0.4)
x_st = np.linspace(1.2, 5.3, 300)
scale2 = max_n / kde2(x_st).max()
ax.plot(x_st, kde2(x_st) * scale2, color='#F6C05C', linewidth=2.5, linestyle='-')

# ── Garis Mean ──
mean_cv = df['CV Score'].mean()
mean_st = df['Stress Score'].mean()
med_cv  = df['CV Score'].median()
med_st  = df['Stress Score'].median()

ax.axvline(mean_cv, color=BIRU,  linewidth=2,   linestyle='--', label=f'Mean CV = {mean_cv:.2f}')
ax.axvline(mean_st, color=EMAS,  linewidth=2,   linestyle='--', label=f'Mean Stres = {mean_st:.2f}')
ax.axvline(med_cv,  color=BIRU_M,linewidth=1.5, linestyle=':',  label=f'Median CV = {med_cv:.2f}')
ax.axvline(med_st,  color='#F6C05C', linewidth=1.5, linestyle=':', label=f'Median Stres = {med_st:.2f}')

# ── Anotasi statistik ──
sd_cv  = df['CV Score'].std()
sd_st  = df['Stress Score'].std()
sk_cv  = df['CV Score'].skew()
sk_st  = df['Stress Score'].skew()

info = (
    f'  CV Score          Stress Score\n'
    f'  Mean : {mean_cv:.2f}          {mean_st:.2f}\n'
    f'  Median: {med_cv:.2f}          {med_st:.2f}\n'
    f'  SD    : {sd_cv:.2f}           {sd_st:.2f}\n'
    f'  Min   : {df["CV Score"].min():.2f}           {df["Stress Score"].min():.2f}\n'
    f'  Max   : {df["CV Score"].max():.2f}           {df["Stress Score"].max():.2f}\n'
    f'  Skew  : {sk_cv:.2f}          {sk_st:.2f}'
)
ax.text(0.015, 0.97, info, transform=ax.transAxes,
        fontsize=8.5, va='top', ha='left', color='#CBD5E0', family='monospace',
        bbox=dict(boxstyle='round,pad=0.6', facecolor='#0D1B2A', alpha=0.88))

# ── Styling ──
ax.set_title(
    'Histogram Distribusi: CV Score vs Stress Score Mahasiswa ITSB (n=36)',
    fontsize=13, fontweight='bold', color='white', pad=12
)
ax.set_xlabel('Skor Rata-rata (Skala Likert 1–5)', color='#CBD5E0', fontsize=11)
ax.set_ylabel('Frekuensi Responden', color='#CBD5E0', fontsize=11)
ax.tick_params(colors='#CBD5E0')
ax.set_xlim(1.2, 5.5)
for spine in ax.spines.values(): spine.set_edgecolor('#2D3748')

ax.legend(
    fontsize=9, facecolor='#1B2A3B', labelcolor='white',
    framealpha=0.9, loc='upper right'
)

plt.tight_layout()
plt.savefig('histogram_overlay.png', dpi=150, bbox_inches='tight', facecolor=BG)
plt.show()
print('Histogram overlay tersimpan sebagai histogram_overlay.png')
```

|-----------|-------|-------|
| **Mean** | ~3.56 | Rata-rata tekanan CV berada di atas titik tengah skala |
| **Median** | ~3.57 | Mendekati mean → distribusi relatif simetris |
| **SD** | ~0.6–0.7 | Variasi antar-responden tergolong sedang |
| **Skewness** | Mendekati 0 | Distribusi tidak terlalu menceng; sebagian besar responden merasakan tekanan CV di kisaran **3–4** |

> **Pola dominan**: Histogram CV Score memperlihatkan konsentrasi frekuensi terbesar di rentang **3.0–4.0**, yang mengonfirmasi bahwa sebagian besar mahasiswa merasakan tekanan CV pada level *tinggi* namun tidak ekstrem. #### Stress Score

| Statistik | Nilai | Makna |
|-----------|-------|-------|
| **Mean** | ~3.50 | Rata-rata stres tepat di atas batas tengah skala |

> - Posisi **Mean > Median** pada Stress Score memperkuat indikasi adanya responden dengan stres tinggi yang menggeser rata-rata ke atas, sementara CV Score menunjukkan distribusi yang lebih **simetris dan merata**.
